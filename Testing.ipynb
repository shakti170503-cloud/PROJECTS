```python

def main():
    def display_menu():
        print("Contact Book Menu:")
        print("""1. Add Contact
2. View Contact
3. Edit Contact
4. Delete Contact
5. List All Contacts
6. Exit  """)
    
    contact_book = {}

    def add_contact():
        name = input()
        phone = input()
        email = input()
        address = input()
        if name in contact_book:
            print("Contact already exists!")
            return
        else:
            contact_book[name] = {
    "phone": phone,
    "email": email,
    "address": address}
        print("Contact added successfully!")

    def view_contact():
        name = input()
        
        if name in contact_book: 
            a = contact_book[name]
            print("Name:", name)
            print("Phone:", a["phone"])
            print("Email:", a["email"])
            print("Address:", a["address"])
        else:
            print("Contact not found!")
        
    def edit_contact():
        name = input()
        if name in contact_book:
            a = contact_book[name]
            phone = input()
            email = input()
            address = input()
            a["phone"] = phone
            a["email"] = email
            a["address"] = address
            contact_book[name] = a
            print("Contact updated successfully!")

        else:
            print("Contact not found!")
        
    def delete_contact():
        name = input()
        if name in contact_book:
            contact_book.pop(name)
            print("Contact deleted successfully!")

        else:
            print("Contact not found!")
        
    def list_all_contacts():
        if len(contact_book) == 0:
            print("No contacts available.")
        else:
            for i in contact_book:
                if i in contact_book:
                    a = contact_book[i]
                    print("Name:", i)
                    print("Phone:", a["phone"])
                    print("Email:", a["email"])
                    print("Address:", a["address"])
                    print()

   
    while True:
        display_menu()
        a = input()
        if a == 1 or a == "1":
            add_contact()
        elif a == 2 or a == "2":
            view_contact()
        elif a == 3 or a == "3":
            edit_contact()
        elif a == 4 or a == "4":
            delete_contact()
        elif a == 5 or a == "5":
            list_all_contacts() 
        elif a == 6 or a == "6":
            break
        else:
            print("Invalid choice. Please try again.") 
           
main()

```

```python

import time
import msvcrt

def timed_input(prompt, seconds=10):
    answer = ""
    deadline = time.monotonic() + seconds

    print(prompt, end="", flush=True)

    while True:
        remaining = deadline - time.monotonic()

        if remaining <= 0:
            print("\nTime's up!")
            return None

        if msvcrt.kbhit():
            char = msvcrt.getwch()

            if char == "\r":  # Enter
                print()
                return answer

            elif char == "\b":  # Backspace
                if answer:
                    answer = answer[:-1]
                    print("\b \b", end="", flush=True)

            elif char in ("\x00", "\xe0"):
                msvcrt.getwch()  # Ignore special keys

            else:
                answer += char
                print(char, end="", flush=True)

        time.sleep(0.02)


questions = [
    ("What is the area of a circle?\nA. πr  B. 2πr  C. πr²  D. πrh", "C"),
    ("What is the capital of Assam?\nA. Guwahati  B. Dispur  C. Dibrugarh  D. Tezpur", "B"),
    ("Which longitude is opposite the Prime Meridian?\nA. Prime Meridian  B. Equator  C. 180° meridian  D. Tropic of Cancer", "C"),
    ("What is the area of a circle?\nA. πr  B. 2πr  C. πr²  D. πrh", "C"),
    ("What is the full form of AI?\nA. Artificial Intelligence  B. Air Inversion  C. Anti Isobar  D. Auto Initiative", "A"),
    ("How many MB are in 2 GB (using 1 GB = 1024 MB)?\nA. 1024  B. 3096  C. 4084  D. 2048", "D"),
    ("Who is your GOAT?\nA. CR7  B. LM10  C. NY11  D. Pele10", "A"),
    ("What is the capital of Hungary?\nA. Budapest  B. Berlin  C. Frankfurt  D. Munich", "A"),
    ("Who is known as the Father of the Indian Constitution?\nA. Mahatma Gandhi  B. Jawaharlal Nehru  C. Indira Gandhi  D. B. R. Ambedkar", "D"),
    ("What is the chemical formula of hydrogen peroxide?\nA. H2O2  B. H2O3  C. HO2  D. H2O", "A"),
]

print("WELCOME TO THIS QUIZ GAME")
print("10 questions | 10 points each | No negative marking")
print("Pass mark: 50 | Time per question: 10 seconds")

ready = input("Are you ready? Enter Y or N: ").strip().upper()
points = 0

if ready == "Y":
    print("Get ready! Starting in 5 seconds...")
    time.sleep(5)

    for number, (question, correct) in enumerate(questions, start=1):
        print(f"\nQuestion {number}/10")
        answer = timed_input(question + "\nYour answer (A/B/C/D): ", 10)

        if answer is not None and answer.strip().upper() == correct:
            print("Correct!")
            points += 10
        else:
            print(f"Incorrect or skipped. Correct answer: {correct}")

    print("\nQUIZ COMPLETED")
    print(f"Your score is: {points}/100")
    print("PASSED" if points >= 50 else "FAILED")

else:
    print("Thank you for coming!")
```
